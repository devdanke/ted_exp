```python
import ipywidgets as widgets
from ipywidgets import interact, interactive, fixed, interact_manual
```

```python
from ipywidgets import Dropdown, IntRangeSlider, VBox, interactive_output
from IPython.display import display


dropdown = Dropdown(options=['foo', 'bar'], value='foo', description='Choice:')
slider = IntRangeSlider(value=[2022, 2028], min=2020, max=2030, step=1, description='Years:')

def show(choice, years):
    start, end = years
    print(f"Dropdown = {choice}")
    print(f"Selected years: {start} — {end}")
    print("Years list:", list(range(start, end + 1)))

out = interactive_output(show, {'choice': dropdown, 'years': slider})
display(VBox([dropdown, slider, out]))
```

```python
from ipywidgets import Output

def show(choice, years):
    start, end = years
    out = Output()
    with out:
        out.clear_output()
        print(f"Dropdown = {choice}")
        print(f"Selected years: {start} — {end}")
        print("Years list:", list(range(start, end + 1)))
    return out
        

dropdown = Dropdown(options=['foo', 'bar'], value='foo', description='Choice:')
slider = IntRangeSlider(value=[2022, 2028], min=2020, max=2030, step=1, description='Years:')

w = interactive(show, choice=dropdown, years=slider)
display(w)
```
