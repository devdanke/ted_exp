```python

# Configuration
suppliers = {
    'nippon-metal': ['ninja-roofing', 'rice-roofers', 'global-roof'],
    'us-steel': ['zztop-roof', 'ranger-roofing', 'global-roof']
}

part_ids = [
    'tin-clip', 'tin-decoration', 'gold-clip', 'gold-decoration',
    'glass-clip', 'glass-decoration', 'roof-polish'
]

start_date = datetime(2023, 3, 1)
end_date = datetime(2026, 1, 31)

# Pricing factors (rough base per unit)
pricing = {
    'tin-clip': 100,
    'tin-decoration': 90,
    'gold-clip': 500,
    'gold-decoration': 400,
    'glass-clip': 1000,
    'glass-decoration': 800,
    'roof-polish': 200
}

# Generate data
rows = []

current_date = start_date
po_counter = 1

while current_date <= end_date:
    year = current_date.year
    month = current_date.month

    # Decide if this month has purchases (10/12 months)
    if random.random() < (10/12):
        for supplier, licensees in suppliers.items():
            for licensee in licensees:
                # Skip some months randomly per licensee
                if random.random() < 0.1:
                    continue

                po = f"PO{po_counter:04d}"
                po_counter += 1

                # 3 to 7 random parts per order
                num_parts = random.randint(3, 7)
                selected_parts = random.sample(part_ids, num_parts)

                order_date = current_date + timedelta(days=random.randint(1, 25))

                for part in selected_parts:
                    qty = random.randint(40, 250)
                    amt = qty * pricing[part] * random.uniform(0.9, 1.3)  # price variation
                    amt = int(round(amt / 100) * 100)  # round to nearest 100

                    rows.append({
                        'DATE': order_date.strftime('%Y-%m-%d'),
                        'SUPPLIER': supplier,
                        'LICENSEE': licensee,
                        'PO': po,
                        'PART_ID': part,
                        'QTY': qty,
                        'AMT': amt
                    })

    # Move to next month
    if month == 12:
        current_date = datetime(year + 1, 1, 1)
    else:
        current_date = datetime(year, month + 1, 1)

# Create DataFrame
df = pd.DataFrame(rows)

# Save to CSV
df.to_csv('sales-grok.csv', index=False)

print(f"Generated {len(df)} rows. Saved to sales-grok.csv")
df.head()
```
